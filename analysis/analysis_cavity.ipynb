```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
import glob
```

```python
files = sorted(glob.glob('../process/temp/*cavity*.fits'))
files
```

```python
cavities = []

for file in files:
    with fits.open(file) as hdul:
        data = hdul[0].data

    cavities.append(data)

cavities = np.array(cavities)
```

```python
plt.figure(figsize=(10,10))
plt.imshow(cavities[1], 'seismic', vmin=-0.07, vmax=0.07)
plt.tight_layout()
```

```python
plt.figure(figsize=(10,10))
plt.imshow(cavities[3] - cavities[2], 'seismic', vmin=-0.01, vmax=0.01)
plt.tight_layout()
```
