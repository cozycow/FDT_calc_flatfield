```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
import glob
from scipy.ndimage import gaussian_filter
```

```python
files = sorted(glob.glob('../process/temp/*cavity*.fits'))
files
```

```python
cavities = []

for file in files:
    with fits.open(file) as hdul:
        data = hdul[0].data

    cavities.append(data)

cavities = np.array(cavities)
```

```python
cavities.shape
```

```python
plt.figure(figsize=(10,10))
plt.imshow(cavities[0,0], 'seismic', vmin=-0.07, vmax=0.07)
plt.tight_layout()
```

```python
plt.figure(figsize=(10,10))
plt.imshow(cavities[1,0] - cavities[0,0], 'seismic', vmin=-0.01, vmax=0.01)
plt.tight_layout()
```

```python
i = 9

temp = cavities[i,0] - cavities[i,1]
temp -= np.nanmean(temp[750:1250,750:1250])
temp = gaussian_filter(temp, 30)

plt.figure(figsize=(10,10))
plt.imshow(temp, 'seismic', vmin=-1e-4, vmax=1e-4)
plt.tight_layout()
```

```python
np.savez('/home/ulyanov/data/solo/bias.npz', bias=temp)
```
