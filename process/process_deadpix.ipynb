```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from processing import *
from scipy.ndimage import gaussian_filter
import glob
```

```python
dark_file = '/home/ulyanov/data/solo/phi/dark/solo_CAL1_phi-fdt-dark_20240205T033810_V202402220119C_0422051001.fits.gz'
```

```python
folders = sorted(glob.glob('/home/ulyanov/data/solo/phi/flat/fdt/calibration/*'))[::-1]
folders
```

```python
folder = folders[3]
files = sorted(glob.glob(folder + '/*.fits.gz'))
mask = np.ones((2048, 2048)).astype(bool)

for file in files:
    data, header = process(file, dark_file=dark_file)
    contpos = header['CONTPOS'] - 1
    mask *= data[contpos,0] < 2e3

mask = ~mask
```

```python
plt.figure(figsize=(10,10))
plt.imshow(mask, 'gray')
plt.tight_layout()
```
