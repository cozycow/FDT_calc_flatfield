```python
folder = folders[3]
files = sorted(glob.glob(folder + '/*.fits.gz'))
mask = np.ones((2048, 2048)).astype(bool)

for file in files:
    data, header = process(file, dark_file=dark_file)
    contpos = header['CONTPOS'] - 1
    mask *= data[contpos,0] < 2e3

mask = ~mask
mask = mask & deadpix
```
