```python
flat_files = sorted(glob.glob('/home/ulyanov/data/solo/phi/flat/temp/*flat*.fits'))
ghost_files = sorted(glob.glob('/home/ulyanov/data/solo/phi/flat/temp/*ghost*.fits'))
flat_dids = [flat_file.split('_')[-1].split('.')[0] for flat_file in flat_files]
ghost_dids = [ghost_file.split('_')[-1].split('.')[0] for ghost_file in ghost_files]
flat_files
```

```python
folders = sorted(glob.glob('/home/ulyanov/data/solo/phi/flat/fdt/calibration/*'))[::-1]
folders
```

```python
for folder in folders[:]:
    files = sorted(glob.glob(folder + '/*.fits.gz'))
    did = files[0].split('_')[-1].split('.')[0]

    flat_file = flat_files[flat_dids.index(did)]
    ghost_file = ghost_files[ghost_dids.index(did)]

    calc_cavity(folder, 'temp',
               dark_file=dark_file,
               deadpix_file=deadpix_file,
               flatfield_file=flat_file,
               ghost_file=ghost_file,
               prefilter_file=prefilter_file,
               distortion_file=distortion_file,
               niter=1000,
               quicklook=False)
```
