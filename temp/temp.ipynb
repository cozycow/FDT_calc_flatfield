```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt

from calc_flatfield import *
import glob
from kll import kll
```

```python
files = sorted(glob.glob('out/*'))
files
```

```python
shifts = []
images = []
centers = []

for file in files:
    with fits.open(file) as hdul:
        header = hdul[0].header
        data = hdul[0].data

    contpos = int(header['CONTPOS']) - 1
    xc = header['CRPIX2'] - 1
    yc = header['CRPIX1'] - 1

    shifts += [get_wv_shift(data, header)]
    images += [data[contpos,0].copy()]
    centers += [(xc, yc)]

centers = np.array(centers)
shifts = np.array(shifts)
images = np.array(images)
```

```python
plt.figure(figsize=(10,10))
plt.imshow(images[4], cmap='gray')
plt.tight_layout()
```

```python
plt.figure(figsize=(10,10))
plt.imshow(shifts[4], cmap='bwr', vmin=-0.05, vmax=0.05)
plt.tight_layout()
```

```python
cavity = kll(shifts, centers, images.clip(0), niter=10, sigma=1e-3, vmin=-0.1, vmax=0.1)
```

```python
plt.figure(figsize=(10,10))
plt.imshow(cavity, cmap='bwr', vmin=-0.05, vmax=0.05)
plt.tight_layout()
```

```python
cavity_ = kll(shifts, np.round(centers).astype(int), images.clip(0), niter=20, sigma=None, vmin=-0.1, vmax=0.1)
```

```python
plt.figure(figsize=(10,10))
plt.imshow(cavity_, cmap='bwr', vmin=-0.05, vmax=0.05)
plt.tight_layout()
```

```python
temp = cavity_ - cavity
temp -= np.nanmedian(temp[900:1100,900:1100])

plt.figure(figsize=(10,10))
plt.imshow(temp, cmap='bwr', vmin=-0.005, vmax=0.005)
plt.tight_layout()
```
